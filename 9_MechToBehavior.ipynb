### **1. The 12-Cell Microcircuit Architecture**
As explained in the Lab Manual, our network is composed of three "classes" of neurons:
* **PN (Principal Cells):** The 8 excitatory output neurons that drive fear behavior.
* **PV (Parvalbumin):** 2 fast-spiking interneurons providing perisomatic "veto" inhibition.
* **SOM (Somatostatin):** 2 dendritic-targeting interneurons that gate synaptic inputs.

### **2. The Hebbian Learning Engine**
The model uses a **Calcium-Controlled Plasticity Rule**. As you run the external code, watch the Rasters closely during the **Conditioning** phase: the simultaneous arrival of the Tone (CS) and Shock (US) triggers the calcium influx necessary to "save" the fear memory.

### 🎯 Design Problem — Apply Inhibitory Timing
**Scenario:** In the 12-cell lateral amygdala network, inhibitory (PV/SOM) onset is shifted to occur *earlier* relative to the excitatory tone input, with no other change.

**What happens to the principal neuron's response peak and latency?**

A) Peak decreases and effective latency increases — earlier inhibition suppresses the response before it can fully build, and delays when it crosses threshold
B) Peak increases and latency decreases — earlier inhibition amplifies excitation
C) Neither peak nor latency changes — inhibitory timing only affects extinction, not acquisition
D) Peak decreases but latency also decreases

*Reason it out before revealing the answer.*

**Answer: A).**

Earlier inhibitory onset means the inhibitory "brake" is already engaged while excitation is still building — this caps how high the response can climb (lower peak) and pushes back the point where the response finally crosses threshold (longer latency). This is exactly the "Gatekeeper" idea from this Set's own reflection questions: inhibitory timing reshapes the response without changing the strength of the tone input itself.

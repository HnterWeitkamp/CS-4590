## A Neuron's Job: Signaling
Every neuron you built in Set 3 — no matter how carefully you assembled its leak channels, its capacitor, its equation — has one job: **signaling**. Before going any further, it's worth asking directly: what does that actually mean?

**Activity — Ask AI:** *"What does it mean for a neuron to 'signal'? What information might it need to communicate, and what would it mean for that signal to arrive quickly versus slowly?"* Paste the response.

For a neuron to “signal” means that it changes its electrical state in a way that can influence another cell. Usually, this involves:

Receiving inputs from other neurons or sensory receptors. Integrating those inputs in its cell body and dendrites. If the combined input is strong enough, generating an action potential—a brief electrical impulse traveling along the axon. At the axon terminal, releasing neurotransmitters that affect the next neuron, muscle, or gland. The signal does not necessarily contain a complete, explicit message like a sentence. Information can be represented through:

Which neurons are active How often they fire—their firing rate The precise timing of spikes The pattern of activity across many neurons Which neurotransmitters and receptors are involved Sometimes the strength or duration of a more gradual electrical signal For example, a sensory neuron might communicate the presence of pressure, its intensity, its location, and how quickly the pressure is changing.

A signal arriving quickly means there is little delay between an event and the neuron’s response. This is important for rapid reactions such as withdrawing from pain, maintaining balance, or coordinating movement. Fast signaling can preserve precise timing, allowing the brain to compare when signals arrive from different locations.

A signal arriving slowly may be less useful for immediate reactions but can still carry important information. Slow signals may reflect gradual changes, such as hormone levels, hunger, inflammation, or sustained attention. They can also produce longer-lasting effects, influencing the neuron’s state over seconds, minutes, or longer.

Thus, neural signaling communicates both what is happening and when it is happening. The speed, timing, frequency, and duration of the signal can all contribute to its meaning

### Recall: How Fast Is What You Already Built?
You already have a working answer to "how does this neuron respond to input" — the passive membrane from Set 3_2. Let's look at it again, honestly, with fresh eyes.

```python
# Quick re-observation: a passive membrane's response to a step of injected current
import numpy as np
import matplotlib.pyplot as plt

C = 1.0       # uF/cm^2 (specific capacitance)
g_leak = 0.1  # mS/cm^2
E_leak = -65  # mV
I = 5.0       # uA/cm^2, injected starting at t=10ms

dt = 0.05
t = np.arange(0, 100, dt)
V = np.full_like(t, E_leak)
for i in range(1, len(t)):
    I_now = I if t[i] > 10 else 0
    dV = (-g_leak*(V[i-1]-E_leak) + I_now) / C
    V[i] = V[i-1] + dV*dt

plt.figure(figsize=(7,4))
plt.plot(t, V)
plt.xlabel('time (ms)'); plt.ylabel('V_m (mV)')
plt.title('The passive membrane you built in Set 3_2, responding to a current step')
plt.show()
print(f'Time constant tau = C/g_leak = {C/g_leak:.1f} ms')
```

**Look at that curve.** It takes several milliseconds just to climb partway to its new resting value — and biological signaling often needs to happen in a fraction of a millisecond. This passive response, on its own, is simply **too slow** to be the whole story.

### Back to the Store
Remember Set 3_2's shopping trip? The bilayer, the leak channels, the ions, the rig? It turns out there's one more aisle we skipped.

Every leak channel you bought was a **fixed resistor** — its resistance never changes, no matter what's happening to the membrane voltage around it. That fixed-ness is exactly what makes the passive response slow: current has to accumulate on the capacitor gradually, through a resistance that never budges to help things along.

What's sitting on the next shelf is a different kind of part entirely: a **voltage-gated channel** — a resistor whose resistance *changes* depending on the voltage across it. Depolarize the membrane a little, and this channel's resistance can drop sharply, letting a rush of current through right when it's needed — a self-reinforcing, much faster response than any fixed leak channel could ever produce.

```python
#@title LIF Experimental Rig { run: "auto" }
import numpy as np
import matplotlib.pyplot as plt

# Parameters from sliders
R = 100 #@param {type:"slider", min:50, max:250, step:5}
tau = 0.02 #@param {type:"slider", min:0.005, max:0.1, step:0.001}
V_rest, V_th, V_reset = -65, -50, -70

t = np.arange(0, 1.0, 1e-4)
I = np.zeros_like(t)
I[(t>0.1)&(t<0.6)] = 0.5; I[(t>0.7)&(t<0.8)] = 0.8
V = np.ones_like(t)*V_rest; spikes = []

for k in range(1, len(t)):
    V[k] = V[k-1] + ((-(V[k-1]-V_rest) + R*I[k-1])*(1e-4/tau))
    if V[k] >= V_th:
        spikes.append(t[k]); V[k] = V[k-1] = V_reset

plt.figure(figsize=(8, 3.5))
plt.plot(t, V, 'k'); plt.ylabel('V (mV)')
plt.twinx(); plt.plot(t, I, 'C0', alpha=0.3); plt.ylabel('I (nA)')
plt.title(f'LIF Response | Spikes: {len(spikes)}')
plt.show()
```

```python
#@title Izhikevich Parameter Tuning { run: "auto" }
import numpy as np
import matplotlib.pyplot as plt

a = 0.02 #@param {type:"slider", min:0.01, max:0.1, step:0.01}
b = 0.2 #@param {type:"slider", min:0.05, max:0.3, step:0.01}
c = -65 #@param {type:"slider", min:-75, max:-45, step:1}
d = 8 #@param {type:"slider", min:0.05, max:10.0, step:0.05}

T = 1000; I = np.zeros(T); I[200:800] = 10
v = -65 * np.ones(T); u = b * v

for k in range(1, T):
    dv = 0.04*v[k-1]**2 + 5*v[k-1] + 140 - u[k-1] + I[k-1]
    du = a*(b*v[k-1] - u[k-1])
    v[k] = v[k-1] + dv; u[k] = u[k-1] + du
    if v[k] >= 30:
        v[k-1]=30; v[k]=c; u[k]+=d

plt.figure(figsize=(8, 3.5)); plt.plot(v, 'k')
plt.title('C2: Interactive Izhikevich'); plt.ylabel('v (mV)'); plt.show()
```

```python
#@title C3: Hodgkin-Huxley Experimental Rig { run: "auto" }
from neuron import h
import numpy as np
import matplotlib.pyplot as plt

# 1. Clear memory first
h('forall delete_section()')

# 2. Parameters from Sliders
stim_amp = 0.1 #@param {type:"slider", min:0.05, max:0.3, step:0.01}
stim_dur = 600 #@param {type:"slider", min:100, max:800, step:50}
stim_delay = 100 #@param {type:"slider", min:10, max:300, step:10}

# 3. Rig Setup
soma = h.Section(name='soma_hh')
soma.L = soma.diam = 12.6157
soma.insert('hh')

stim = h.IClamp(soma(0.5))
stim.delay = stim_delay
stim.dur = stim_dur
stim.amp = stim_amp

# 4. Recording & Run
v = h.Vector().record(soma(0.5)._ref_v)
t = h.Vector().record(h._ref_t)
h.finitialize(-65)
h.continuerun(800)

# 5. Plotting
plt.figure(figsize=(8, 4))
plt.plot(t, v, 'k', lw=1.5, label='Membrane Voltage')
plt.axhline(-65, color='gray', linestyle='--', alpha=0.5)
plt.axvline(stim_delay, color='C0', linestyle=':', alpha=0.8, label='Stimulus Onset')
plt.title(f'HH Current Clamp | Delay: {stim_delay}ms | Amp: {stim_amp}nA')
plt.xlabel('Time (ms)'); plt.ylabel('Membrane Voltage (mV)')
plt.legend(loc='upper right'); plt.grid(alpha=0.2)
plt.show()
```

```python
#@title C4: Sensor-to-Soma Data Bridge { run: "auto" }
data_scale = 0.005 #@param {type:"slider", min:0.001, max:0.02, step:0.001}
data_offset = 0.05 #@param {type:"slider", min:-0.1, max:0.2, step:0.01}

from neuron import h
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a dummy "Real-world" sensor signal (e.g., an Accelerometer)
t_data = np.linspace(0, 500, 1000)
raw_sensor = 50 + 30 * np.sin(2 * np.pi * 0.01 * t_data) + 10 * np.random.randn(1000)

# 2. Map Sensor -> Model Amps
# We apply the user's scaling and offset here
mapped_amps = (raw_sensor * data_scale) + data_offset

# 3. Setup NEURON Rig
h('forall delete_section()')
soma_c4 = h.Section(name='somac4')
soma_c4.L = soma_c4.diam = 20
soma_c4.insert('hh')

# Use play() to drive the injection with the mapped array
stim_vec = h.Vector(mapped_amps)
t_vec_stim = h.Vector(t_data)
stim = h.IClamp(soma_c4(0.5))
stim.delay = 0
stim.dur = 1e9 # Keep open to follow the vector
stim_vec.play(stim._ref_amp, t_vec_stim, 1)

# Record
v = h.Vector().record(soma_c4(0.5)._ref_v)
t = h.Vector().record(h._ref_t)

h.finitialize(-65)
h.continuerun(500)

# 4. Plotting
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

ax1.plot(t_data, raw_sensor, color='gray', alpha=0.5)
ax1.set_ylabel('Raw Sensor Units')
ax1.set_title('External Data Input')

ax2.plot(t, v, 'k')
ax2.set_ylabel('Membrane V (mV)')
ax2.set_xlabel('Time (ms)')
ax2.axhline(-50, color='red', ls=':', alpha=0.5, label='Threshold')

plt.tight_layout()
plt.show()
```

### 🎯 Design Problem — Parameter Transparency
**Scenario:** Two models produce identical-looking spike trains for the same input. Model A uses parameters in real biophysical units (mV, nA, ms) — e.g., $g_{Na}=0.12$ S/cm². Model B uses unitless, arbitrary "strength" scores (e.g., "channel strength = 7.3") tuned until its output visually matches Model A's.

**Which model is more scientifically useful, and why?**

A) Model A — biophysical units let you interpret, validate, and compare results against real recordings; Model B's fit could be right for the wrong reasons
B) Model B — fewer constraints means it's more flexible and therefore more useful
C) They're equally useful, since both produce the same output
D) Model A is worse, since real units make the model harder to tune

*Reason it out before revealing the answer.*

**Answer: A).**

Matching output alone doesn't mean matching mechanism. Model B's arbitrary units could be compensating for a completely wrong underlying assumption — there's no way to tell, because there's nothing to check the numbers against. Model A's biophysical units mean every parameter can be independently validated against real measurements (e.g., does $g_{Na}=0.12$ match a real recorded sodium conductance?), and every result can be meaningfully compared to other studies.

### 🚫 Hand-Work — No AI for This Part
The **derivative** of f(x) at a point is the slope of the tangent line there (limit definition):
$$f'(x) = \lim_{\Delta x \to 0} \frac{f(x+\Delta x) - f(x)}{\Delta x}$$
Every differentiation formula you'll ever use comes from this one definition.

**Worked example — f(x) = 2x:**
$$f'(x) = \lim_{\Delta x \to 0} \frac{2(x+\Delta x) - 2x}{\Delta x} = \lim_{\Delta x \to 0} \frac{2\Delta x}{\Delta x} = 2$$

A **first-order** DE contains only first derivatives ($\dot y$, not $\ddot y$). We'll use a general pattern covering a wide range of systems:
$$a_1 \dot y(t) + a_2\, y(t) = a_3\, u(t), \quad y(0) = y_0$$
which can always be rewritten in **standard form** by dividing through by $a_2$:
$$\tau \dot y(t) + y(t) = K\, u(t), \quad y(0) = y_0$$
where $\tau = a_1/a_2$ is the **time constant** (how fast the system responds) and $K = a_3/a_2$ is the **gain** (where it settles). $u(t)$ is the driving input, often a constant.

Here are two very different-looking systems that are secretly the same math.

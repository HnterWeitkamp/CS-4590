Short answer

Yes — in a limited, useful sense a neuron is like a battery: both create and maintain an electrical potential difference that can do work.
But the neuron is not a single passive electrochemical cell with a fixed voltage; it is a dynamic, distributed, actively maintained system whose voltage can change rapidly and locally.
How a neuron’s electrical signaling is actually powered

Ionic concentration gradients: neurons maintain different concentrations of key ions (mainly K+, Na+, Cl−, and Ca2+) across the plasma membrane. Typical resting concentrations: high K+ inside, high Na+ and Ca2+ outside.
Electrogenic pumps and transporters: membrane proteins (most notably the Na+/K+ ATPase) use metabolic energy (ATP) to move ions against their gradients. The Na+/K+ pump exchanges 3 Na+ out for 2 K+ in, which is electrogenic (contributes a small net negative charge inside).
Resting membrane potential: the combination of concentration gradients and ion-selective permeability (chiefly K+ permeability at rest) produces a membrane potential (typically ~ −60 to −75 mV in many neurons).
Voltage- and ligand-gated ion channels: signaling (action potentials, synaptic potentials) comes from transient changes in membrane conductance through ion channels. When channels open, ions flow down their electrochemical gradients, producing rapid voltage changes.
Capacitive and resistive properties: the membrane behaves electrically like a capacitor (stores charge) in parallel with variable resistances (ion channels). Charging/discharging underlies the time course of potentials.
Ways a neuron is similar to a battery

Maintains a potential difference: both produce a steady voltage between two terminals (inside vs outside for a neuron).
Energy source is chemical: batteries and neurons both convert chemical energy into an electrochemical potential. In neurons the chemical energy comes from ATP-driven pumps (biochemical reactions); in batteries from redox chemistry.
Can drive current: when conductance/pathway is provided, the stored energy drives ion flow (current) that can do work (e.g., change membrane potential, trigger transmitter release).
Key differences from a simple battery

Not a fixed voltage source: a battery has approximately a fixed open-circuit voltage dictated by its chemistry. A neuron's membrane potential is set by multiple ions and permeabilities and is dynamic — it can change rapidly (milliseconds) and locally (on a small patch of membrane).
Variable conductance and regenerative channels: neurons use voltage-gated channels that change conductance in a nonlinear, state-dependent way. This allows regenerative events (action potentials) — positive feedback not present in an ordinary battery circuit.
Distributed RC system vs single cell: the neuron is spatially extended and behaves as a distributed cable (resistances and capacitances vary along dendrites and axons). A battery is a lumped two-terminal device.
Non-equilibrium steady state: the neuron maintains ionic gradients by continuous ATP expenditure; the gradients would dissipate if pumps stopped. A battery also eventually exhausts, but neuronal gradients are actively maintained and tightly coupled to metabolism.
Multiple ion species and reversal potentials: the effective voltage arises from several ions. The Nernst equation gives the equilibrium potential for a single ion; the Goldman equation / Goldman-Hodgkin-Katz equation gives the membrane potential when multiple permeant ions contribute. A battery typically has a single well-defined EMF.
Nernst example (for ion X): E_X = (RT/zF) ln([X]_out/[X]_in)
Resting Vm often close to E_K because membrane is most permeable to K+ at rest.
Small voltages, small charge movement: the membrane potential is on the order of tens of millivolts (typically −70 mV), and tiny numbers of ions moving cause large potential changes because the membrane capacitance is small. That differs from macroscopic batteries where large numbers of electrons move.
Energy and direction depend on gradients: during an action potential, Na+ influx depolarizes the membrane because [Na+] is high outside; after the AP, pumps must restore the gradients, using ATP.
Useful analogies and the electrical circuit model

Equivalent circuit used in neuroscience: membrane capacitance (Cm) in parallel with ion-specific conductances (g_K, g_Na, …) and battery-like sources (E_K, E_Na, …). The batteries in the model are not physical single cells but represent Nernst potentials due to concentration gradients.
Membrane current equation (simplified): Cm dV/dt = −g_K (V − E_K) − g_Na (V − E_Na) − g_leak (V − E_leak) + I_injected
The Hodgkin–Huxley model formalized this and explains action potentials as time- and voltage-dependent changes in conductances.
Practical implications

Because ion pumps use ATP, neuronal signaling is metabolically costly — the brain consumes a lot of energy to maintain ionic gradients.
Pathologies that disrupt pumps or ion gradients (ischemia, toxins) rapidly impair signaling.
The battery analogy is useful pedagogically (there is a voltage produced by chemical gradients) but misses key dynamic features: variable conductance, regeneration, metabolic dependence, and spatial complexity.

### The Answer: An Electrochemical Gradient
A neuron's "battery" isn't a chemical reaction driving electrons through a wire — it's a **concentration difference** of ions (mainly K⁺ and Na⁺) held across the membrane, the same membrane you'll model as a capacitor starting in Set 3. K⁺ is concentrated *inside* the cell; Na⁺ is concentrated *outside*. Separating charge like this, across an insulator, stores potential energy — exactly the way a battery does, just with ion gradients as the "fuel" instead of a chemical reaction.

Short answer

The Na+/K+ ATPase (sodium–potassium pump) is the main active mechanism that restores Na+ and K+ gradients after activity. Other active transporters (Ca2+ pumps, SERCA, PMCA), exchangers (NCX), and secondary active transporters also contribute for specific ions and compartments.
The Na+/K+ pump uses chemical energy from ATP hydrolysis: each cycle hydrolyzes 1 ATP to export 3 Na+ and import 2 K+ (electrogenic). Restoring gradients therefore consumes substantial ATP.
Rough energetic share: maintaining and restoring ionic gradients (chiefly via the Na+/K+ pump) is one of the largest energy sinks in neurons — typically on the order of tens of percent of a neuron's ATP use at rest. For brain gray matter a common estimate is ~40–60% of ATP consumption goes to Na+/K+ ATPase activity (exact numbers vary by cell type, activity level, and method).
How the main players work

Na+/K+ ATPase (primary active transport)
Stoichiometry: 3 Na+ out : 2 K+ in per ATP hydrolyzed.
Mechanism (simplified): pump binds 3 intracellular Na+, ATP phosphorylates the pump → conformational change exposes Na+ to outside and releases them; then binds 2 extracellular K+, pump is dephosphorylated → returns to original conformation and releases K+ inside.
Functional result: restores Na+ and K+ concentration gradients after activity and contributes a small net outward current (pump is electrogenic).
Ca2+ removal
Plasma membrane Ca2+ ATPase (PMCA) and sarco/endoplasmic reticulum Ca2+ ATPase (SERCA) use ATP directly to pump Ca2+ out of the cytosol into extracellular space or intracellular stores.
Na+/Ca2+ exchanger (NCX) often uses the Na+ gradient (secondary active transport) to export Ca2+ without direct ATP use, but maintaining that Na+ gradient indirectly costs ATP (via Na+/K+ ATPase).
Cotransporters and exchangers
Various Na+-coupled transporters (e.g., Na+/K+/2Cl− cotransporter) move other ions or transmitters and are ultimately powered by the Na+ gradient that the pump maintains.
What it costs energetically

Per cycle energy: 1 ATP ≈ 20 kBT biochemical energy available under cellular conditions (≈ −50 kJ/mol free energy depending on concentrations/conditions). Practically, each Na+/K+ ATPase cycle uses one ATP to move 3 Na+ and 2 K+.
Translating to ion bookkeeping:
To remove N Na+ that entered during activity, you need ≈ N/3 ATP (ignoring other flux pathways and exchanger contributions).
Restoring a single action potential (AP) to baseline: literature estimates vary with cell type and AP shape, but typical cortical axonal/dendritic estimates put the ATP cost per AP in the range 10^6–10^9 ATP molecules depending on how many ions crossed (myelination, axon diameter, firing pattern). The large range reflects cell morphology and how broadly you count presynaptic vs postsynaptic costs.
Fraction of total energy budget at rest
Multiple studies (classically Attwell & Laughlin 2001 and follow-ups) estimate that a large fraction of neuronal ATP at baseline powers ion pumps. Reasonable, commonly cited ballpark for gray-matter neurons:
Na+/K+ ATPase (maintaining/restoring Na+/K+ gradients): ~40–60% of total ATP consumption in many preparations.
Additional ATP for Ca2+ handling, vesicle recycling, protein synthesis, and housekeeping consumes the remainder.
Important caveats: these percentages vary substantially with cell type, temperature, species, and especially activity level. In very quiet/resting neurons the fraction might be lower; in highly active neurons a larger share of total brain ATP goes to restoring gradients after spiking and synaptic currents.
Why this is metabolically important

The pump continuously runs to oppose leak and activity-driven influxes. If ATP supply fails (ischemia, metabolic poison), pumps stop, ionic gradients collapse within seconds–minutes, membrane potential depolarizes, and signaling and ion homeostasis fail — a major mechanism for ischemic neuronal injury.
Because of the large ATP demand, a substantial fraction of brain glucose/oxygen consumption supports these ion-pumping processes.

### Quantifying the Battery: The Nernst Equation
How much voltage does a given gradient actually store? That's what the **Nernst equation** answers — for a single ion species, it gives the membrane voltage at which that ion's electrical and chemical forces exactly balance:

$$E_{ion} = \frac{RT}{zF} \ln\left(\frac{[\text{ion}]_{outside}}{[\text{ion}]_{inside}}\right)$$

where $R$ is the gas constant, $T$ is temperature (Kelvin), $z$ is the ion's charge (+1 for K⁺), and $F$ is Faraday's constant. Don't worry about memorizing this — you're about to have AI help you turn it into code.

### Activity 3 — Vibe-Code the Nernst Equation
Let's compute a real, historically famous number: the potassium reversal potential of the **squid giant axon** — the preparation Hodgkin and Huxley used to first work out how neurons spike. Typical values: $[K^+]_{outside} \approx 20$ mM, $[K^+]_{inside} \approx 400$ mM, at $T \approx 279.5$ K ($6.3°C$, their experimental temperature).

> **Suggested prompt:** *"Write Python code that implements the Nernst equation as a function `nernst(K_out, K_in, T=279.5, z=1)`, using the gas constant R=8.314 and Faraday's constant F=96485, returning the result in millivolts. Then use it to compute the potassium reversal potential for the squid giant axon with K_out=20 mM and K_in=400 mM."*

Paste and run the code below.

```python
import math

R = 8.314      # J / (mol·K)  — universal gas constant
F = 96485      # C / mol      — Faraday's constant


def nernst(K_out, K_in, T=279.5, z=1):
    """
    Nernst equilibrium (reversal) potential for a single ion species.

        E_ion = (R*T)/(z*F) * ln([ion]_out / [ion]_in)

    Parameters
    ----------
    K_out : float
        Extracellular ion concentration (any unit, e.g. mM).
    K_in : float
        Intracellular ion concentration (same unit as K_out).
    T : float, optional
        Absolute temperature in kelvin (default 279.5 K ≈ 6.35 °C,
        a typical squid-axon experimental temperature).
    z : int, optional
        Valence of the ion (+1 for K+, +2 for Ca2+, -1 for Cl-).

    Returns
    -------
    float
        Reversal potential in millivolts (mV).

    Notes
    -----
    Only the *ratio* of concentrations matters, so units cancel as long
    as K_out and K_in are expressed in the same units.
    """
    if K_out <= 0 or K_in <= 0:
        raise ValueError("Concentrations must be positive.")
    if z == 0:
        raise ValueError("Valence z must be non-zero.")

    E_volts = (R * T) / (z * F) * math.log(K_out / K_in)
    return E_volts * 1000.0          # V -> mV


if __name__ == "__main__":
    # Squid giant axon, resting potassium gradient
    K_out, K_in = 20.0, 400.0        # mM

    E_K = nernst(K_out, K_in)

    print(f"Thermal voltage RT/F = {R * 279.5 / F * 1000:.3f} mV")
    print(f"[K+]out/[K+]in       = {K_out/K_in:.3f}")
    print(f"E_K                  = {E_K:.2f} mV")
```

### Activity 4 — Build Your Own Interactive Nernst "Experiment"
A single number is fine, but an experiment lets you *see* how the gradient controls the voltage. Let's build sliders for the inside and outside concentrations, so you can drag them and watch the reversal potential respond live — no re-running cells, no re-typing numbers.

> **Suggested prompt:** *"Using ipywidgets in Google Colab, build an interactive tool with two sliders — one for [K+]outside (range 5 to 50 mM) and one for [K+]inside (range 100 to 500 mM) — that recomputes and displays the Nernst potential live as I move either slider, using the nernst() function from before. Show the result as printed text in millivolts, updating automatically."*

Paste and run the code below. Play with both sliders before moving on.

```python
import math
from ipywidgets import FloatSlider, VBox, interactive_output, Label
from IPython.display import display

# Nernst function: returns potential in volts
def nernst(outside, inside, T_celsius=37.0, z=1.0):
    """
    Compute the Nernst potential E (volts) for an ion with charge z
    given outside and inside concentrations (same units).
    Default: K+ (z=+1) at 37 °C.
    """
    R = 8.314462618  # J/(mol*K)
    F = 96485.33212  # C/mol
    T = T_celsius + 273.15
    return (R * T / (z * F)) * math.log(float(outside) / float(inside))

# Sliders: ranges per your request
k_out_slider = FloatSlider(value=5.0, min=5.0, max=50.0, step=0.1,
                           description='[K+] outside (mM):', continuous_update=True)
k_in_slider  = FloatSlider(value=140.0, min=100.0, max=500.0, step=0.1,
                           description='[K+] inside (mM):', continuous_update=True)

# Output label to show mV
output_label = Label()

def update(k_out, k_in):
    E_volts = nernst(k_out, k_in, T_celsius=37.0, z=1.0)
    E_mV = E_volts * 1000.0
    output_label.value = f"Nernst potential (K+): {E_mV:.2f} mV  (T=37 °C, z=+1)"

# Wire widgets to output and display
out = interactive_output(update, {'k_out': k_out_slider, 'k_in': k_in_slider})
display(VBox([k_out_slider, k_in_slider, output_label]))
```

## Pillar 5 — How Circuits Learn: Hebb's Rule
Two traits are hardwired into essentially every animal species, from insects to humans: **survival and reproduction**. One way to survive is *innate* fear — some fears you're simply born with. The other is the ability to **learn fear fast**.

Think about touching a hot stove for the first time. You don't need a lesson, a textbook, or a second try — one bad experience, and it's locked into memory. That's not a coincidence; it reflects a circuit-level rule for how connections between neurons change:

> **Hebb's Rule (Donald O. Hebb, 1949): "Cells that fire together, wire together."**

When two neurons are active at the same time, again and again, the connection between them tends to strengthen. This one idea is the conceptual seed for two things later in this course: the calcium-based plasticity rule you'll work with directly in **Set 8**, and the very circuit you're about to reverse-engineer below.

### The Three Phases
